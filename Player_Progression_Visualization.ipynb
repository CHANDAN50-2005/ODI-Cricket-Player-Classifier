# Player Progression Visualization Demo

This notebook demonstrates how to visualize a cricket player's performance progression over time using ODI data. We'll plot runs scored per year for a selected player.

## 1. Import Required Libraries
We'll use pandas for data handling and matplotlib/plotly for visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
```

## 2. Load and Preview ODI Data
We'll load the ODI data and preview the first few rows.

```python
# Load ODI data
odi_df = pd.read_csv('ODI data.csv')
odi_df.head()
```

## 3. Select a Player and Prepare Progression Data
Choose a player and group their runs by year.

```python
# Example: Select a player by name
player_name = 'Virat Kohli'  # Change to any player in your dataset

# Check available columns
print(odi_df.columns)

# Group runs by year for the selected player
player_df = odi_df[odi_df['Player'] == player_name]
yearly_runs = player_df.groupby('Year')['Runs_Scored'].sum().reset_index()
yearly_runs
```

## 4. Visualize Player Progression
Plot runs scored per year using matplotlib and plotly.

```python
# Matplotlib line chart
plt.figure(figsize=(10,5))
plt.plot(yearly_runs['Year'], yearly_runs['Runs_Scored'], marker='o', color='navy')
plt.title(f'{player_name} - Runs Scored Per Year')
plt.xlabel('Year')
plt.ylabel('Runs Scored')
plt.grid(True)
plt.show()

# Plotly interactive chart
fig = px.line(yearly_runs, x='Year', y='Runs_Scored', title=f'{player_name} - Runs Scored Per Year', markers=True)
fig.show()
```

## 5. Wickets and Batting Average Progression
Let's visualize wickets taken per year and batting average per year for the selected player.

```python
# Wickets per year
if 'Wickets_Taken' in player_df.columns:
    yearly_wickets = player_df.groupby('Year')['Wickets_Taken'].sum().reset_index()
    plt.figure(figsize=(10,5))
    plt.plot(yearly_wickets['Year'], yearly_wickets['Wickets_Taken'], marker='o', color='darkgreen')
    plt.title(f'{player_name} - Wickets Taken Per Year')
    plt.xlabel('Year')
    plt.ylabel('Wickets Taken')
    plt.grid(True)
    plt.show()
    fig = px.line(yearly_wickets, x='Year', y='Wickets_Taken', title=f'{player_name} - Wickets Taken Per Year', markers=True)
    fig.show()
else:
    print('Wickets_Taken column not found in data.')

# Batting average per year
if 'Batting_Average' in player_df.columns:
    yearly_avg = player_df.groupby('Year')['Batting_Average'].mean().reset_index()
    plt.figure(figsize=(10,5))
    plt.plot(yearly_avg['Year'], yearly_avg['Batting_Average'], marker='o', color='orange')
    plt.title(f'{player_name} - Batting Average Per Year')
    plt.xlabel('Year')
    plt.ylabel('Batting Average')
    plt.grid(True)
    plt.show()
    fig = px.line(yearly_avg, x='Year', y='Batting_Average', title=f'{player_name} - Batting Average Per Year', markers=True)
    fig.show()
else:
    print('Batting_Average column not found in data.')
```
