```python
# --- Enhanced Categorization Function with Debug Logging ---
def get_category_final(row, debug=False):
    """
    Categorize cricket players with improved rules and debug logging
    """
    # Extract all stats with clear names
    matches = float(row.get('Mat_x', row.get('Matches', row.get('Mat', 0)))) or 0
    runs = float(row.get('Runs_x', row.get('Runs', 0))) or 0
    bat_avg = float(row.get('Ave_x', row.get('Ave', 0))) or 0
    strike_rate = float(row.get('SR_x', row.get('SR', 0))) or 0
    centuries = float(row.get('100', 0)) or 0
    fifties = float(row.get('50', 0)) or 0
    wickets = float(row.get('Wkts', 0)) or 0
    bowl_avg = float(row.get('Ave_y', 0)) or 0
    economy = float(row.get('Econ', 0)) or 0
    catches = float(row.get('Ct', 0)) or 0
    stumpings = float(row.get('St', 0)) or 0
    
    if debug:
        print(f"\nDEBUG - Player Stats:")
        print(f"Matches: {matches}")
        print(f"Runs: {runs}")
        print(f"Batting Average: {bat_avg}")
        print(f"Strike Rate: {strike_rate}")
        print(f"Centuries: {centuries}")
        print(f"Fifties: {fifties}")
        print(f"Wickets: {wickets}")
        print(f"Bowling Average: {bowl_avg}")
        print(f"Economy: {economy}")
        print(f"Catches: {catches}")
        print(f"Stumpings: {stumpings}")
    
    # Professional Criteria (Strong thresholds)
    if matches >= 50:  # Minimum matches threshold
        # Batting Excellence
        batting_points = 0
        if bat_avg >= 40: batting_points += 2
        if strike_rate >= 85: batting_points += 1
        if runs >= 4000: batting_points += 2
        if centuries >= 10: batting_points += 2
        if fifties >= 25: batting_points += 1
        
        # Bowling Excellence
        bowling_points = 0
        if wickets >= 100: bowling_points += 2
        if economy > 0 and economy < 5.0: bowling_points += 2
        if bowl_avg > 0 and bowl_avg < 35: bowling_points += 2
        
        # Fielding Excellence
        fielding_points = 0
        if catches >= 100: fielding_points += 2
        if stumpings >= 50: fielding_points += 2
        
        if debug:
            print(f"\nDEBUG - Points:")
            print(f"Batting Points: {batting_points}")
            print(f"Bowling Points: {bowling_points}")
            print(f"Fielding Points: {fielding_points}")
            print(f"Total Points: {batting_points + bowling_points + fielding_points}")
        
        # Professional if:
        # 1. Great batting (4+ batting points) OR
        # 2. Great bowling (4+ bowling points) OR
        # 3. Good all-round performance (5+ total points) OR
        # 4. Clear statistical excellence
        is_prof_batsman = batting_points >= 4
        is_prof_bowler = bowling_points >= 4
        is_prof_fielder = fielding_points >= 3
        is_prof_allrounder = (batting_points + bowling_points + fielding_points) >= 5
        is_statistical_great = (runs > 5000 and bat_avg > 40) or (wickets > 150 and bowl_avg < 35)
        
        if debug:
            print(f"\nDEBUG - Professional Criteria:")
            print(f"Is Professional Batsman: {is_prof_batsman}")
            print(f"Is Professional Bowler: {is_prof_bowler}")
            print(f"Is Professional Fielder: {is_prof_fielder}")
            print(f"Is Professional All-rounder: {is_prof_allrounder}")
            print(f"Is Statistical Great: {is_statistical_great}")
        
        if is_prof_batsman or is_prof_bowler or is_prof_fielder or is_prof_allrounder or is_statistical_great:
            return 'Professional'
    
    # Beginner Criteria (More strict)
    if matches < 30 or (matches < 50 and (runs < 1000 and wickets < 30)):
        return 'Beginner'
    
    # Everyone else is Intermediate
    return 'Intermediate'
```

```python
import pandas as pd

# Load the three datasets into pandas DataFrames
batting_df = pd.read_csv('ODI data.csv')
bowling_df = pd.read_csv('Bowling_ODI.csv')
fielding_df = pd.read_csv('Fielding_ODI.csv')

# Let's print the first few rows of each file to make sure they loaded correctly
print("--- Batting Data Loaded ---")
display(batting_df.head())

print("\n--- Bowling Data Loaded ---")
display(bowling_df.head())

print("\n--- Fielding Data Loaded ---")
display(fielding_df.head())
```

```python
# --- Inspect the original Batting Data ---
print("--- Original Batting Data (First 5 Rows) ---")
display(batting_df.head())

# --- Clean the Batting Data ---
# 1. Clean Player names by removing text in parentheses
batting_df['Player'] = batting_df['Player'].str.extract(r'([A-Za-z\s\.\-]+)')

# 2. Clean 'HS' (High Score) column by removing the '*'
batting_df['HS'] = batting_df['HS'].astype(str).str.replace('*', '', regex=False)

# 3. Identify all columns that should be numbers
numeric_cols = ['Mat', 'Inns', 'NO', 'Runs', 'HS', 'Ave', 'BF', 'SR', '100', '50', '0']

# 4. Convert these columns to numbers, changing any errors (like '-') to an empty value
for col in numeric_cols:
    batting_df[col] = pd.to_numeric(batting_df[col], errors='coerce')

# 5. Drop the useless 'Unnamed: 13' column if it exists
if 'Unnamed: 13' in batting_df.columns:
    batting_df = batting_df.drop('Unnamed: 13', axis=1)

# 6. Fill any empty cells that were created with 0
batting_df = batting_df.fillna(0)


# --- Verify the Cleanup ---
print("\n--- Cleaned Batting Data (First 5 Rows) ---")
display(batting_df.head())
print("✅ Batting data has been successfully cleaned.")
```

```python
# --- Inspect the original Fielding Data ---
print("--- Original Fielding Data (First 5 Rows) ---")
display(fielding_df.head())

# --- Clean the Fielding Data ---
# 1. Clean Player names
fielding_df['Player'] = fielding_df['Player'].str.extract(r'([A-Za-z\s\.\-]+)')

# 2. Identify columns that should be numbers
numeric_cols = ['Mat', 'Inns', 'Dis', 'Ct', 'St', 'Ct Wk', 'Ct Fi']

# 3. Convert columns to numbers, changing errors to NaN
for col in numeric_cols:
    fielding_df[col] = pd.to_numeric(fielding_df[col], errors='coerce')

# 4. Drop the useless column if it exists
if 'Unnamed: 11' in fielding_df.columns:
    fielding_df = fielding_df.drop('Unnamed: 11', axis=1)

# 5. Fill all empty numeric cells with 0
fielding_df = fielding_df.fillna(0)

# --- Verify the Cleanup ---
print("\n--- Cleaned Fielding Data (First 5 Rows) ---")
display(fielding_df.head())
print("✅ Fielding data has been successfully cleaned.")
```

```python
# --- Merge the Cleaned DataFrames ---

# 1. Merge the batting and bowling data first
# We use an 'outer' merge to keep all players from both tables.
master_df = pd.merge(batting_df, bowling_df, on=['Player', 'Span'], how='outer')

# 2. Now, merge the result with the fielding data
master_df = pd.merge(master_df, fielding_df, on=['Player', 'Span'], how='outer')

# 3. The merge might create duplicate 'Matches' columns (Matches_x, Matches_y).
# Let's clean that up by keeping the main one.
if 'Matches_x' in master_df.columns:
    master_df.rename(columns={'Matches_x': 'Matches'}, inplace=True)
    if 'Matches_y' in master_df.columns:
        master_df = master_df.drop('Matches_y', axis=1)

# 4. Final cleanup: Fill any empty cells that were created during the merge with 0
master_df = master_df.fillna(0)


# --- Verify the Final Master DataFrame ---
print("--- Our Final Master Player Dataset ---")
print(f"The master table has {master_df.shape[0]} rows and {master_df.shape[1]} columns.")
display(master_df.head())
```

```python
import pandas as pd
import numpy as np

print("--- Final Step: Merging Data and Creating Correct Categories ---")

# --- Part 1: Merge the three clean datasets ---
master_df = pd.merge(batting_df, bowling_df, on=['Player', 'Span'], how='outer')
master_df = pd.merge(master_df, fielding_df, on=['Player', 'Span'], how='outer')
master_df.fillna(0, inplace=True)

# --- Part 2: Apply the improved categorization ---
# Apply the function to each row
print("\nCategorizing players...")
master_df['Category'] = master_df.apply(lambda row: get_category_final(row, debug=False), axis=1)

# Show the distribution of categories
print("\n--- Category Distribution ---")
category_counts = master_df['Category'].value_counts()
print("\nPlayer counts per category:")
print(category_counts)
print("\nPercentages:")
print((category_counts / len(master_df) * 100).round(1), "%")

# Show some example professionals
print("\n--- Example Professional Players ---")
pros = master_df[master_df['Category'] == 'Professional'].head()
print("\nProfessional Players Sample:")
display(pros[['Player', 'Mat_x', 'Runs_x', 'Ave_x', 'SR_x', '100', '50', 'Wkts']])
```

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

print("--- Preparing Data for Feature Scaling (Final Corrected Version) ---")

# --- 1. Separate the features (X) from the target (y) ---
# We select all columns EXCEPT the ones that are not stats.
columns_to_drop = ['Player', 'Span', 'Category', 'BBI', 'HS']
X = master_df.drop(columns=columns_to_drop, errors='ignore')
y = master_df['Category']
print("✅ Features (X) and Target (y) have been separated.")

# --- 2. THE FINAL FIX: A safety check to ensure all columns in X are numbers ---
# We loop through each column in our feature set (X).
# If any column is still text ('object'), we force it to become a number.
# This prevents the scaler from crashing.
for col in X.columns:
    if X[col].dtype == 'object':
        X[col] = pd.to_numeric(X[col], errors='coerce').fillna(0)
print("✅ Safety check complete: All feature columns are now guaranteed to be numeric.")


# --- 3. Create and Apply the Scaler ---
# This will now work because X is guaranteed to be clean.
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

print("✅ All player stats have been successfully scaled.")
print("Our data is now perfectly balanced and ready for training!")
```

```python
from sklearn.model_selection import train_test_split

print("--- Splitting the data into training and testing sets ---")

# We use the scaled data (X_scaled) to create our sets.
# 80% will be for training, 20% for testing.
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# --- Verify the split ---
print("✅ Data has been successfully split.")
print(f"Shape of our training clues (X_train): {X_train.shape}")
print(f"Shape of our testing clues (X_test):  {X_test.shape}")
```

```python
from sklearn.neighbors import KNeighborsClassifier

print("--- Training the AI Model ---")

# 1. Create our kNN model and tell it to look at the 5 nearest "neighbors"
knn_final = KNeighborsClassifier(n_neighbors=5)

# 2. Train the model using our perfectly scaled training data
knn_final.fit(X_train, y_train)

print("✅ The AI model has been successfully trained!")
```

```python
from sklearn.metrics import accuracy_score, classification_report

print("--- Evaluating the Final, Smartest Model ---")

# 1. Ask the trained model to make predictions on the test data
y_pred = knn_final.predict(X_test)

# 2. Calculate the accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"\n--- Final Model Accuracy ---")
print(f"Our model is {accuracy * 100:.2f}% accurate!")
print("This means it correctly predicted the category for that percentage of players in the test set.")

# 3. Display the detailed report card
print("\n--- Classification Report ---")
print(classification_report(y_test, y_pred))
```
